# Pandas

### In this notebook, we will learn about Sorting and Basic Statistics in Pandas.

* Here, I am importing all the required dependencies.

```python
import numpy as np
import pandas as pd
```

## 1. Sorting Data

* Here, we will learn how to sort a DataFrame by single or multiple columns using `sort_values()` and by row indices using `sort_index()`.

```python
# Here, I am creating a new DataFrame from a Python dictionary and printing it.

data = {
    "Product": ["Pen", "Notebook", "Pencil", "Marker", "Eraser", "Sharpener"],
    "Price": [10, 50, 5, 30, 8, 12],
    "Stock": [100, 60, 200, 80, 150, 120],
    "Rating": [4.5, 4.7, 4.2, 4.8, 4.3, 4.4]
}

df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
```

```python
# Here, I am sorting the DataFrame by a single column ('Price', ascending) and printing it.

print("\nSorted by Price (ascending):")
print(df.sort_values(by="Price"))
```

```python
# Here, I am sorting the DataFrame by a single column ('Rating', descending) and printing it.

print("\nSorted by Rating (descending):")
print(df.sort_values(by="Rating", ascending=False))
```

```python
# Here, I am sorting by multiple columns ('Price' ascending, then 'Stock' descending) and printing it.

print("\nSorted by Price (asc) and Stock (desc):")
print(df.sort_values(by=["Price", "Stock"], ascending=[True, False]))

# 💡 Pro-Tip: When sorting by multiple columns, the second column ('Stock') acts as a tie-breaker 
# whenever two or more rows have the exact same value in the first column ('Price')!
```

```python
# Here, I am sorting the DataFrame by index (row labels) in descending order and printing it.

print("\nSorted by index (descending):")
print(df.sort_index(ascending=False))
```

## 2. Descriptive Statistics

* Here, we will compute key summary statistics (`mean`, `median`, `mode`, `std`, and `describe`) across the DataFrame.

```python
# Here, I am finding the mean (average) of all numeric columns and printing it.

print("\nMean of numeric columns:")
print(df.mean(numeric_only=True))
```

```python
# Here, I am finding the median (middle value) of all numeric columns and printing it.

print("\nMedian of numeric columns:")
print(df.median(numeric_only=True))
```

```python
# Here, I am finding the mode (most frequent value) of each column and printing it.

print("\nMode of each column:")
print(df.mode())

# 💡 Pro-Tip: Since every value in our small dataset appears only once (frequency = 1), 
# Pandas treats all values as tied for the mode and returns all 6 rows sorted!
```

```python
# Here, I am finding the standard deviation of all numeric columns and printing it.

print("\nStandard deviation of numeric columns:")
print(df.std(numeric_only=True))
```

```python
# Here, I am finding the full descriptive statistical summary of the whole DataFrame and printing it.

print("\nDescriptive statistics summary:")
print(df.describe())

# 💡 Pro-Tip: By default, df.describe() only summarizes numeric columns. 
# Use df.describe(include="all") to also see summary stats (count, unique, top, freq) for text columns!
```

## 3. Counting Values

* Here, we will calculate the frequency and percentage distribution of unique values in a column using `value_counts()`.

```python
# Here, I am counting how many times each product appears in the DataFrame and printing it.

print("\nFrequency of each product:")
print(df["Product"].value_counts())
```

```python
# Here, I am finding the relative proportions (percentages) instead of raw counts for each product and printing it.

print("\nPercentage frequency of each product:")
print(df["Product"].value_counts(normalize=True))

# 💡 Pro-Tip: Multiply by 100 (e.g., df["Product"].value_counts(normalize=True) * 100) to view values directly as percentages (0–100%)!
```

```python
# Here, I am finding the frequency of each rating and printing it.

print("\nFrequency of each rating:")
print(df["Rating"].value_counts())
```

## 4. Correlation & Covariance

* Here, we will measure the linear relationship and directional variance between numeric columns using `corr()` and `cov()`.

```python
# Here, I am finding the correlation matrix between all numeric columns and printing it.

print("\nCorrelation matrix:")
print(df.corr(numeric_only=True))

# 💡 Pro-Tip: Correlation ranges from -1 to +1. Notice how 'Price' and 'Stock' have a strong negative correlation (-0.80), meaning higher-priced items tend to have lower stock!
```

```python
# Here, I am finding the covariance matrix between all numeric columns and printing it.

print("\nCovariance matrix:")
print(df.cov(numeric_only=True))
```

```python
# Here, I am finding the correlation between two specific columns ('Price' and 'Stock') and printing it.

print("\nCorrelation between Price and Stock:")
print(df["Price"].corr(df["Stock"]))
```

```python
# Here, I am finding the covariance between two specific columns ('Price' and 'Stock') and printing it.

print("\nCovariance between Price and Stock:")
print(df["Price"].cov(df["Stock"]))
```
