## 1. Converting Columns to Datetime 

* Here, I am converting strings or numbers into datetime objects using `pd.to_datetime()`.

```python
# Here, I am creating a sample DataFrame with date strings and printing it.

data = {
    "OrderID": [1001, 1002, 1003, 1004, 1005],
    "OrderDate": ["2024-01-05", "2024/01/12", "15-01-2024", "2024-01-20", "20240125"],
    "Customer": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Amount": [250, 400, 150, 300, 500]
}

df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
```

```python
# Here, I am converting the 'OrderDate' column to datetime objects using pd.to_datetime().

df["OrderDate"] = pd.to_datetime(df["OrderDate"], dayfirst=True, errors="coerce")
print("\nAfter converting OrderDate to datetime:")
print(df)
```

```python
# Here, I am verifying the data types to confirm the conversion to datetime64.

print("\nData types after conversion:")
print(df.dtypes)
```

```python
# Here, I am handling rows where conversion failed.

if df["OrderDate"].isna().any():
    print("\nRows with invalid dates:")
    print(df[df["OrderDate"].isna()])
```
