# Pandas

### In this notebook, we will learn about Data Cleaning in Pandas.

* Here, I am importing all the required dependencies.

```python
import numpy as np
import pandas as pd
```

## 1. Handling Missing Values

* Here, we will learn how to detect, count, drop, and fill missing values (`None` / `NaN`) in a DataFrame.

```python
# Here, I am creating a DataFrame with some missing values using a Python dictionary and printing it.

a = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Age": [25, None, 35, 40, None],
    "City": ["Delhi", "Mumbai", None, "Bangalore", "Chennai"],
    "Score": [85, 91, np.nan, 88, 95]
}

df = pd.DataFrame(a)
print("\nOriginal DataFrame:")
print(df)

# 💡 Pro-Tip: Notice how integer columns with missing values ('Age' and 'Score') automatically become float64 because NaN is a floating-point value under the hood.
```

```python
# Here, I am detecting missing values across the DataFrame.

print("\nCheck where data is missing:")
print(df.isna())

# 💡 Pro-Tip: df.isnull() is an exact alias for df.isna() in Pandas.
```

```python
# Here, I am printing the total count of missing values in each column.

print("\nCount of missing values in each column:")
print(df.isna().sum())
```

```python
# Here, I am dropping rows that contain any missing values and saving the result into another variable.

c = df.dropna()
print("\nAfter dropping rows with missing values:")
print(c)

# 💡 Pro-Tip: Use df.dropna(subset=["Age"]) if you only want to drop rows where a specific column is missing!
```

```python
# Here, I am copying the DataFrame into another variable and filling the missing values.
# Filling numeric columns with the mean or a constant, and text columns with a placeholder.

d = df.copy()
d["Age"] = d["Age"].fillna(d["Age"].mean())
d["City"] = d["City"].fillna("Unknown")
d["Score"] = d["Score"].fillna(0)

print("\nAfter filling missing values:")
print(d)

# 💡 Pro-Tip: Reassigning the column (d["Age"] = d["Age"].fillna(...)) is the modern Pandas 2.0+ best practice instead of using inplace=True on a column!
```

## 2. Changing Data Types

* Here, we will see how to convert column data types using the `astype()` method.

```python
# Here, I am creating a DataFrame using a Python dictionary and printing it.

a = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Age": [25, 30, 35, 40, 22],
    "Score": [88, 92, 79, 85, 95]
}

b = pd.DataFrame(a)
print("Original DataFrame:")
print(b)
```

```python
# Here, I am converting the "Age" column from int to float and checking the data types.

b["Age"] = b["Age"].astype(float)
print("\n'Age' column converted to float:")
print(b.dtypes)
```

```python
# Here, I am converting the "Name" column to category type (memory-efficient) and checking the data types.

b["Name"] = b["Name"].astype("category")
print("\n'Name' column converted to category:")
print(b.dtypes)

# 💡 Pro-Tip: Converting low-cardinality text columns (like Gender, City, or Department) to 'category' drastically reduces RAM usage on large datasets!
```

```python
# Here, I am converting the "Score" column to string and checking the data types.

b["Score"] = b["Score"].astype(str)
print("\n'Score' column converted to string:")
print(b.dtypes)
```

## 3. Replacing Values

* Here, we will learn how to replace single values, multiple values, and string patterns in a DataFrame.

```python
# Here, I am replacing a single value in a column.

c = df.copy()
c["Name"] = c["Name"].replace("Bob", "Robert")
print("\nAfter replacing 'Bob' with 'Robert':")
print(c)
```

```python
# Here, I am replacing multiple values at once in a column.

d = b.copy()
d["Score"] = d["Score"].astype(int).replace({88: 90, 95: 99})
print("\nAfter replacing 88 -> 90 and 95 -> 99 in 'Score':")
print(d)
```

```python
# Here, I am replacing values in strings (e.g., shortening names).

e = df.copy()
e["Name"] = e["Name"].replace({"Charlie": "Ch"}, regex=False)
print("\nAfter shortening 'Charlie' to 'Ch':")
print(e)

# 💡 Pro-Tip: Set regex=True if you want to match and replace partial string patterns using Regular Expressions!
```

## 4. Dropping Rows and Columns

* Here, we will see how to remove specific rows and columns from a DataFrame using `drop()`.

```python
# Here, I am dropping a single row by its index label.

c = b.drop(2)  # drops row with index 2 (Charlie)
print("\nAfter dropping row with index 2:")
print(c)
```

```python
# Here, I am dropping multiple rows by passing a list of index labels.

d = b.drop([0, 4])  # drops Alice and Emma
print("\nAfter dropping rows 0 and 4:")
print(d)
```

```python
# Here, I am dropping a single column using axis=1.

e = b.drop("Score", axis=1)
print("\nAfter dropping the 'Score' column:")
print(e)

# 💡 Pro-Tip: Instead of 'axis=1', you can also write b.drop(columns="Score") for better readability!
```

```python
# Here, I am dropping multiple columns at once.

f = b.drop(["Age", "Score"], axis=1)
print("\nAfter dropping 'Age' and 'Score' columns:")
print(f)
```

```python
# Here, I am using inplace=True to modify the DataFrame directly without reassignment.

g = b.copy()
g.drop("Score", axis=1, inplace=True)
print("\nAfter dropping 'Score' with inplace=True:")
print(g)
```

## 5. Detecting & Removing Duplicates

* Here, we will learn how to identify, count, and remove duplicate rows in a DataFrame.

```python
# Here, I am creating a custom dataset with some duplicate rows.

data = {
    "Name": ["Alice", "Bob", "Charlie", "Alice", "Bob", "David"],
    "Age": [25, 30, 35, 25, 30, 40],
    "Score": [88, 92, 79, 88, 92, 85]
}

df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
```

```python
# Here, I am detecting duplicate rows in the DataFrame.

print("\nBoolean Series showing duplicates:")
print(df.duplicated())
```

```python
# Here, I am counting the total number of duplicate rows in the DataFrame.

print("\nTotal number of duplicate rows:")
print(df.duplicated().sum())
```

```python
# Here, I am removing duplicate rows from the DataFrame (keeping the first occurrence).

df_first = df.drop_duplicates()
print("\nAfter dropping duplicates (keep='first'):")
print(df_first)
```

```python
# Here, I am removing duplicate rows from the DataFrame (keeping the last occurrence).

df_last = df.drop_duplicates(keep="last")
print("\nAfter dropping duplicates (keep='last'):")
print(df_last)
```

```python
# Here, I am removing all duplicate rows from the DataFrame (keep=False).

df_none = df.drop_duplicates(keep=False)
print("\nAfter removing all duplicates (keep=False):")
print(df_none)
```

```python
# Here, I am removing duplicate rows based on a specific subset of columns (e.g., 'Name' only).

df_subset = df.drop_duplicates(subset=["Name"])
print("\nAfter dropping duplicates based on 'Name' only:")
print(df_subset)
```
