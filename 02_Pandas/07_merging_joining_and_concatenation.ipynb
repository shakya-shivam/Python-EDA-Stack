# Pandas

* Here, I am importing all the required dependencies.

```python
import numpy as np
import pandas as pd
```

## 1. Using `concat()` for Stacking Data

* Here, we will learn how to stack DataFrames vertically (row-wise, axis=0) and horizontally (column-wise, axis=1) using `pd.concat()`.

```python
# Here, I am creating three sample DataFrames to demonstrate concatenation and printing them.

# Dataset 1: Product details
df1 = pd.DataFrame({
    "ProductID": [101, 102, 103],
    "ProductName": ["Pen", "Notebook", "Marker"],
    "Price": [10, 50, 30]
})

# Dataset 2: More products (same columns as df1)
df2 = pd.DataFrame({
    "ProductID": [104, 105],
    "ProductName": ["Pencil", "Eraser"],
    "Price": [5, 8]
})

# Dataset 3: Additional info (different columns)
df3 = pd.DataFrame({
    "Stock": [100, 60, 80],
    "Rating": [4.5, 4.7, 4.8]
})

print("Original DataFrames:")
print("\nDF1:\n", df1)
print("\nDF2:\n", df2)
print("\nDF3:\n", df3)
```

```python
# Here, I am vertically stacking rows (combining df1 and df2) and printing the result.

vertical = pd.concat([df1, df2], ignore_index=True)
print("\nVertical concatenation of df1 & df2:")
print(vertical)

# 💡 Pro-Tip: Setting ignore_index=True resets the row index to a clean 0, 1, 2, 3, 4 sequence instead of keeping duplicate indices (0, 1, 2, 0, 1)!
```

```python
# Here, I am horizontally stacking DataFrames (combining df1 and df3 column-wise) using axis=1.

horizontal = pd.concat([df1, df3], axis=1)
print("\nHorizontal concatenation of df1 & df3:")
print(horizontal)
```

```python
# Here, I am horizontally stacking DataFrames while preserving the original index alignment.
# (Leaving ignore_index=False keeps the original column names and aligns rows by their index labels).

horizontal_no_reset = pd.concat([df1, df3], axis=1, ignore_index=False)
print("\nHorizontal concat without resetting index:")
print(horizontal_no_reset)
```

## 2. Using `merge()` for SQL-Style Joins

* Here, we will combine DataFrames based on common key columns using SQL-style joins (`inner`, `left`, `right`, and `outer`).

```python
# Here, I am preparing the 'products' and 'sales' DataFrames to demonstrate SQL-style joins.

# Combine vertically so we have one complete product table
products = pd.concat([df1, df2], ignore_index=True)

# Sales data ('ProductID' is the common key for joining)
sales = pd.DataFrame({
    "ProductID": [101, 102, 104, 106],  # 106 doesn't exist in products
    "UnitsSold": [500, 200, 150, 50]
})

print("Products DataFrame:")
print(products)
print("\nSales DataFrame:")
print(sales)
```

```python
# Here, I am performing an inner join to keep only the rows with matching 'ProductID' in both DataFrames.

inner = pd.merge(products, sales, on="ProductID", how="inner")
print("\nInner Join (only matches):")
print(inner)

# 💡 Pro-Tip: If the key columns have different names in the two tables, use left_on="prod_id" and right_on="item_id" instead of on="ProductID"!
```

```python
# Here, I am performing a left join to keep all products and add matching sales data (filling NaN where no match exists).

left = pd.merge(products, sales, on="ProductID", how="left")
print("\nLeft Join (all products):")
print(left)
```

```python
# Here, I am performing a right join to keep all sales records and add matching product info (filling NaN where no match exists).

right = pd.merge(products, sales, on="ProductID", how="right")
print("\nRight Join (all sales):")
print(right)
```

```python
# Here, I am performing a full outer join to keep all records from both DataFrames, filling unmatched values with NaN.

outer = pd.merge(products, sales, on="ProductID", how="outer")
print("\nOuter Join (all data):")
print(outer)

# 💡 Pro-Tip: Pass indicator=True inside pd.merge() to add a '_merge' column that shows whether each row came from 'left_only', 'right_only', or 'both'!
```

## 3. Using `join()` with Index Alignment

* Here, we will use the `.join()` method, which is a convenient shorthand for combining DataFrames based on their row indices.

```python
# Here, I am preparing the 'products' table and an index-aligned 'stock' DataFrame.

# Combine products into one table
products = pd.concat([df1, df2], ignore_index=True)

# Stock info (separate DataFrame, indexed by ProductID)
stock = pd.DataFrame({
    "Stock": [100, 60, 80, 120, 150]
}, index=[101, 102, 103, 104, 105])  # ProductID as index

print("Products DataFrame:")
print(products)
print("\nStock DataFrame (indexed by ProductID):")
print(stock)
```

```python
# Here, I am joining the 'stock' DataFrame with 'products' by aligning both on the 'ProductID' index.
# First, setting 'ProductID' as the index in 'products'.

p_indexed = products.set_index("ProductID")
joined = p_indexed.join(stock)
print("\nJoined DataFrame (products + stock):")
print(joined)
```

```python
# Here, I am using join() with the 'on' parameter so we don't have to set 'ProductID' as the index on the left DataFrame first.

joined_on = products.join(stock, on="ProductID")
print("\nJoined using `on='ProductID'`:")
print(joined_on)

# 💡 Pro-Tip: Use pd.merge() when joining on columns, and use df.join() when at least one DataFrame is already indexed by the join key!
```

## 4. Practical Examples

* Here, we will solve a real-world business scenario by merging 'sales' and 'revenue' datasets that have partially mismatched ProductID keys.

```python
# Here, I am creating sample 'sales' and 'revenue' DataFrames for a practical merging scenario.

# Sales data
sales = pd.DataFrame({
    "ProductID": [201, 202, 203, 204],
    "UnitsSold": [500, 300, 400, 250]
})

# Revenue data (missing ProductID 204, extra ProductID 205)
revenue = pd.DataFrame({
    "ProductID": [201, 202, 203, 205],
    "Revenue": [7500, 4500, 6800, 1200]
})

print("Sales Data:")
print(sales)
print("\nRevenue Data:")
print(revenue)
```

```python
# Here, I am merging 'sales' and 'revenue' using an outer join to see complete product performance and printing the result.

merged_perf = pd.merge(sales, revenue, on="ProductID", how="outer")
print("\nMerged Sales & Revenue (outer join):")
print(merged_perf)
```
