# Pandas

## In this file we will learn about advanced tricks in Pandas.

* Here, I am importing all required dependencies.

```python
import pandas as pd
```

### 1. Applying Custom Functions

```python
# Here, I am creating a sample DataFrame containing student marks.

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Math": [88, 75, 92, 60, 85],
    "Science": [90, 80, 78, 70, 95]
}

df = pd.DataFrame(data)
df
```

```python
# Here, I am calculating the row-wise total marks for each student using apply(axis=1) and a lambda function.

df["Total"] = df[["Math", "Science"]].apply(lambda row: row.sum(), axis=1)
print("\nTotal marks using apply(axis=1):")
print(df)
```

```python
# Here, I am calculating the column-wise mean for numeric columns using apply(axis=0).

col_mean = df[["Math", "Science"]].apply(lambda col: col.mean())
print("\nColumn means using apply(axis=0):")
print(col_mean)
```

```python
# Here, I am applying a custom Python function to convert numeric Math scores into letter grades.

def to_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    else:
        return "D"

df["MathGrade"] = df["Math"].apply(to_grade)
print("\nGrades for Math using apply():")
print(df[["Name", "Math", "MathGrade"]])

# 💡 Pro-Tip: apply() works on entire DataFrames (row-wise or column-wise) AND Series, whereas map() works strictly on a single Series (element-wise).
```

## 2. Using map() for Series Transformations

* Here, we will use the map() method, which is specifically designed for element-wise transformations on a single Pandas Series (column).

```python
# Here, I am converting all student names to lowercase using the map() method.

df["NameLower"] = df["Name"].map(str.lower)
print("\nLowercase names using map():")
print(df[["Name", "NameLower"]])
```

```python
# Here, I am adding 5 bonus marks to the Math column using map() with a lambda function.

df["MathBonus"] = df["Math"].map(lambda x: x + 5)
print("\nMath marks with +5 bonus using map():")
print(df[["Name", "Math", "MathBonus"]])

# 💡 Pro-Tip: While apply() and map() with lambdas are highly flexible, they act like hidden loops and can be slow on large datasets.
#  Whenever possible, use Pandas built-in vectorized operations (e.g., df["Math"] + 5) for exponentially faster performance!
```
