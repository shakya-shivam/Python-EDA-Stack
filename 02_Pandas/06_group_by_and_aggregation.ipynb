# Pandas

### In this notebook, we will learn about GroupBy and Aggregation in Pandas.

* Here, I am importing all the required dependencies.

```python
import numpy as np
import pandas as pd
```

## 1. Concept of Split–Apply–Combine

* Here, we will understand the core **Split–Apply–Combine** paradigm:
  1. **Split:** Divide the data into groups based on some criteria (e.g., `Department`).
  2. **Apply:** Compute an aggregation function on each group independently (e.g., `mean`, `count`).
  3. **Combine:** Merge the results back into a clean summary Series or DataFrame.

```python
# Here, I am creating a new DataFrame with Employee, Department, and Salary data and printing it.

data = {
    "Employee": ["John", "Sarah", "Mike", "Anna", "Tom", "Laura", "Steve", "Kate"],
    "Department": ["HR", "IT", "IT", "Finance", "Finance", "HR", "IT", "Finance"],
    "Salary": [50000, 70000, 72000, 65000, 60000, 52000, 75000, 58000]
}

df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
```

```python
# Here, I am finding the average salary by department and printing it.

avg_salary = df.groupby("Department")["Salary"].mean()
print("\nAverage salary by Department:")
print(avg_salary)

# 💡 Pro-Tip: Use df.groupby("Department", as_index=False)["Salary"].mean() if you want 'Department' to remain a regular column instead of becoming the index!
```

```python
# Here, I am counting the number of employees per department and printing it.

emp_count = df.groupby("Department")["Employee"].count()
print("\nNumber of employees by Department:")
print(emp_count)

# 💡 Pro-Tip: .count() ignores missing (NaN) values, whereas df.groupby("Department").size() counts all rows including NaN!
```

```python
# Here, I am grouping by department and aggregating multiple statistics at once, then printing the result.

multi_stats = df.groupby("Department")["Salary"].agg(["mean", "max", "min"])
print("\nSalary stats (mean, max, min) by Department:")
print(multi_stats)
```

## 2. Using `groupby()` with Aggregation Functions

* Here, we will apply common aggregation functions (`mean()`, `sum()`, `count()`, and `agg()`) directly on grouped data.

```python
# Here, I am finding the mean salary per department and printing it.

print("\nMean salary by Department:")
print(df.groupby("Department")["Salary"].mean())
```

```python
# Here, I am finding the total sum of salaries per department and printing it.

print("\nTotal salary by Department:")
print(df.groupby("Department")["Salary"].sum())
```

```python
# Here, I am counting how many employees are in each department and printing it.

print("\nNumber of employees by Department:")
print(df.groupby("Department")["Employee"].count())
```

```python
# Here, I am performing multiple aggregations on 'Salary' using 'agg()' and printing the result.

print("\nMultiple aggregations on Salary:")
print(df.groupby("Department")["Salary"].agg(["mean", "max", "min", "count"]))
```

## 3. Multiple Aggregations on Different Columns

* Here, we will pass a dictionary to `agg()` to apply different aggregation functions to different columns simultaneously.

```python
# Here, I am creating a DataFrame with an additional 'Age' column using a Python dictionary and printing it.

data = {
    "Employee": ["John", "Sarah", "Mike", "Anna", "Tom", "Laura", "Steve", "Kate"],
    "Department": ["HR", "IT", "IT", "Finance", "Finance", "HR", "IT", "Finance"],
    "Salary": [50000, 70000, 72000, 65000, 60000, 52000, 75000, 58000],
    "Age": [28, 32, 30, 27, 35, 29, 31, 26]
}

df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)

# Here, I am performing multiple aggregations on different columns and printing the result.

result = df.groupby("Department").agg({
    "Salary": ["mean", "max", "min"],
    "Age": ["mean", "max"]
})

print("\nMultiple aggregations on Salary and Age by Department:")
print(result)
```

## 4. Pivot Tables & Crosstab

* Here, we will create Excel-style summary tables using `pivot_table()` and frequency cross-tabulations using `pd.crosstab()`.

```python
# Here, I am calculating the average salary by Department using a pivot table and printing the result.

pivot_salary = df.pivot_table(values="Salary", index="Department", aggfunc="mean")
print("\nPivot table: Average Salary by Department:")
print(pivot_salary)

# 💡 Pro-Tip: Pass 'margins=True' inside pivot_table() to automatically add an 'All' (Grand Total / Overall Average) row!
```

```python
# Here, I am creating a pivot table with multiple values ('Salary' & 'Age') and printing the result.

pivot_multi = df.pivot_table(values=["Salary", "Age"], index="Department", aggfunc="mean")
print("\nPivot table: Average Salary & Age by Department:")
print(pivot_multi)
```

```python
# Here, I am creating an 'AgeGroup' column using 'pd.cut()' and generating a cross-tabulation of Department vs. AgeGroup.

df["AgeGroup"] = pd.cut(df["Age"], bins=[25, 30, 35, 40], labels=["25-30", "31-35", "36-40"])
crosstab_result = pd.crosstab(df["Department"], df["AgeGroup"])

print("\nCrosstab: Employees by Department and Age Group:")
print(crosstab_result)

# 💡 Pro-Tip: Notice that the '36-40' column is hidden because no employee has an age > 35. 
# Pass dropna=False inside pd.crosstab() if you want to display columns with 0 counts too!
```
