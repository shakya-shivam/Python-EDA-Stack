# Pandas

### In this notebook, we will learn about DataFrame basics.

* Here, I am importing all the required dependencies.

```python
import numpy as np
import pandas as pd
```

## 1. Inspecting Data

* Here, we will see how to inspect a DataFrame using built-in methods and attributes.

```python
# Here, I am creating a DataFrame using a Python dictionary and printing it.

a = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Age": [25, 30, 35, 40, 22],
    "Score": [88, 92, 79, 85, 95]
}

b = pd.DataFrame(a)

print(b)
```

```python
# Here, I am printing the first 3 rows using 'head()'.

print(f"\nFirst 3 rows:\n{b.head(3)}")

# 💡 Pro-Tip: By default, b.head() returns the first 5 rows if no number is passed.
```

```python
# Here, I am printing the last 2 rows using 'tail()'.

print(f"\nLast 2 rows:\n{b.tail(2)}")

# 💡 Pro-Tip: By default, b.tail() returns the last 5 rows if no number is passed.
```

```python
# Here, I am printing a concise summary of the DataFrame using 'info()'.

b.info()

# 💡 Pro-Tip: Call b.info() directly without wrapping it inside print(), 
# because info() prints directly to the console and returns 'None'.
```

```python
# Here, I am printing the shape of the DataFrame (rows, columns) using 'shape'.

print(f"\nShape:\n{b.shape}")
```

```python
# Here, I am printing the data types of each column using 'dtypes'.

print(f"\nData types:\n{b.dtypes}")
```

## 2. Selecting Columns and Rows

* Here, we will learn how to select specific columns and rows from a DataFrame.

```python
# Here, I am printing a single column from the DataFrame.

print("Single column (Name):")
print(b["Name"])

# 💡 Pro-Tip: Selecting a single column (b["Name"]) returns a 1D Pandas Series.
```

```python
# Here, I am printing multiple columns from the DataFrame.

print("\nMultiple columns (Name & Score):")
print(b[["Name", "Score"]])

# 💡 Pro-Tip: Passing a list of columns inside double brackets (b[["Name", "Score"]]) returns a 2D DataFrame.
```

```python
# Here, I am printing a specific value from the DataFrame using 'loc' (label-based indexing).

print("\nLabel-based selection (row index 2, column 'Score'):")
print(b.loc[2, "Score"])
```

```python
# Here, I am printing multiple selected rows and columns from the DataFrame using 'loc'.

print("\nRows 1 to 3 with columns Name & Age:")
print(b.loc[1:3, ["Name", "Age"]])

# 💡 Pro-Tip: Slicing with 'loc' (1:3) is INCLUSIVE of both the start (1) and stop (3) labels!
```

```python
# Here, I am printing a specific row from the DataFrame using 'iloc' (integer-location based indexing).

print("\nRow at index 3 (all columns):")
print(b.iloc[3])
```

```python
# Here, I am printing multiple rows and columns from the DataFrame using 'iloc'.

print("\nRows 0 to 2, columns 0 to 1:")
print(b.iloc[0:3, 0:2])

# 💡 Pro-Tip: Unlike 'loc', slicing with 'iloc' (0:3) is EXCLUSIVE of the stop index (3), just like standard Python lists!
```

## 3. Index and Column Overview

* Here, we will inspect the index, column labels, and underlying NumPy array of the DataFrame.


```python
# Here, I am checking the row index labels.

print("Index:")
print(b.index)
```

```python
# Here, I am checking the column names.

print("\nColumns:")
print(b.columns)
```

```python
# Here, I am converting the DataFrame into a NumPy array.

print("\nValues:")
print(b.values)

# 💡 Pro-Tip: In modern Pandas, using 'b.to_numpy()' is officially recommended over 'b.values'.
```

```python
# Here, I am checking the shape of the underlying NumPy array.

print("\nShape of values:")
print(b.values.shape)
```

## 4. Renaming Columns and Setting the Index

* Here, we will see how to rename columns, set a custom column as the index, and reset the index.

```python
# Here, I am renaming a column name.

c = b.rename(columns={"Score": "Marks"})
print("After renaming 'Score' to 'Marks':")
print(c)
```

```python
# Here, I am setting a column as the index.

d = c.set_index("Name")
print("\nAfter setting 'Name' as index:")
print(d)
```

```python
# Here, I am resetting the index back to the default integer index.

e = d.reset_index()
print("\nAfter resetting the index:")
print(e)

# 💡 Pro-Tip: Use d.reset_index(drop=True) if you want to discard the old index instead of adding it back as a column.
```
