# Pandas

### In this notebook, we will learn about Data Selection and Filtering in Pandas.

* Here, I am importing all the required dependencies.

```python
import numpy as np
import pandas as pd
```

## 1. Boolean Indexing and Conditional Selection

* Here, we will learn how to filter rows in a DataFrame using boolean conditions and logical operators (`&`, `|`, `~`).

```python
# Here, I am creating a DataFrame using a Python dictionary and printing it.

a = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Age": [25, 30, 35, 40, 22],
    "Score": [88, 92, 79, 85, 95]
}

b = pd.DataFrame(a)
print("\nOriginal DataFrame:")
print(b)
```

```python
# Here, I am printing the rows where 'Age' is greater than 30.

print("\nRows where Age > 30:")
print(b[b["Age"] > 30])
```

```python
# Here, I am printing the rows where 'Score' is less than 90.

print("\nRows where Score < 90:")
print(b[b["Score"] < 90])
```

```python
# Here, I am applying multiple conditions at the same time using the 'AND' (&) operator.

print("\nRows where Age > 25 AND Score > 85:")
print(b[(b["Age"] > 25) & (b["Score"] > 85)])

# 💡 Pro-Tip: In Pandas, always use '&' instead of 'and', and wrap each condition in parentheses () due to Python's operator precedence rules!
```

```python
# Here, I am applying multiple conditions at the same time using the 'OR' (|) operator.

print("\nRows where Age < 25 OR Score > 90:")
print(b[(b["Age"] < 25) | (b["Score"] > 90)])
```

```python
# Here, I am reversing the condition using the 'NOT' (~) operator.

print("\nRows where Score is NOT greater than 85:")
print(b[~(b["Score"] > 85)])
```

## 2. Filtering Rows with Multiple Conditions

* Here, we will practice combining multiple relational conditions and range filters on a DataFrame.

```python
# Here, I am filtering rows where 'Age' > 25 and 'Score' > 85.

print("\nRows where Age > 25 AND Score > 85:")
print(b[(b["Age"] > 25) & (b["Score"] > 85)])
```

```python
# Here, I am filtering rows where 'Age' < 35 or 'Score' >= 90.

print("\nRows where Age < 35 OR Score >= 90:")
print(b[(b["Age"] < 35) | (b["Score"] >= 90)])
```

```python
# Here, I am filtering rows where 'Age' is between 25 and 35 (inclusive).

print("\nRows where Age is between 25 and 35 (inclusive):")
print(b[(b["Age"] >= 25) & (b["Age"] <= 35)])

# 💡 Pro-Tip: You can write range filters much more cleanly using the built-in .between() method:
# print(b[b["Age"].between(25, 35)])
```

## 3. Using `query()` for Cleaner Filtering

* Here, we will use the `query()` method to filter DataFrames using clean, SQL-like string expressions.

```python
# Here, I am filtering rows where 'Age' > 25 using the 'query()' method.

print("\nRows where Age > 25:")
print(b.query("Age > 25"))
```

```python
# Here, I am filtering rows with multiple conditions ('Age' < 35 and 'Score' >= 85) using 'query()'.

print("\nRows where Age < 35 and Score >= 85:")
print(b.query("Age < 35 and Score >= 85"))

# 💡 Pro-Tip: Inside query() strings, you can directly use English keywords like 'and', 'or', and 'not' instead of '&', '|', and '~'!
```

```python
# Here, I am referencing an external Python variable inside 'query()' using the '@' symbol.

threshold = 90
print(f"\nRows where Score > {threshold}:")
print(b.query("Score > @threshold"))
```

## 4. String Operations on Columns

* Here, we will perform vectorized string operations on text columns using the `.str` accessor.

```python
# Here, I am converting the 'Name' column to lowercase and printing it.

print("\nNames in lowercase:")
print(b["Name"].str.lower())
```

```python
# Here, I am converting the 'Name' column to uppercase and printing it.

print("\nNames in uppercase:")
print(b["Name"].str.upper())
```

```python
# Here, I am filtering rows where 'Name' contains the letter 'a' (case-insensitive) and printing them.

print("\nRows where Name contains 'a':")
print(b[b["Name"].str.contains("a", case=False)])

# 💡 Pro-Tip: In real-world messy datasets with missing values (NaN), always add 'na=False' 
# (e.g., b["Name"].str.contains("a", case=False, na=False)) to prevent ValueError crashes!
```

```python
# Here, I am replacing a substring (changing 'Charlie' to 'Charles') and printing the updated DataFrame.

c = b.copy()
c["Name"] = c["Name"].str.replace("Charlie", "Charles", regex=False)
print("\nAfter replacing 'Charlie' with 'Charles':")
print(c)
```

```python
# Here, I am filtering rows where 'Name' starts with 'D' and printing them.

print("\nRows where Name starts with 'D':")
print(b[b["Name"].str.startswith("D")])
```

```python
# Here, I am filtering rows where 'Name' ends with 'e' and printing them.

print("\nRows where Name ends with 'e':")
print(b[b["Name"].str.endswith("e")])
```
