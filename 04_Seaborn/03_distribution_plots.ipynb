# Seaborn

## Here, we will learn about Distribution Plots in Seaborn, which help us visualize how data points are spread across a continuous range.

* We will start by importing the required dependencies.

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Here, I am loading the built-in 'tips' dataset to practice distribution plots.

tips = sns.load_dataset('tips')
tips
```

### Here, we will create a Histogram using sns.histplot() to see the frequency distribution of the total bill.

```python
# Here, I am creating a basic histogram with 20 bins to group the 'total_bill' values.

sns.histplot(tips["total_bill"], bins=20, color="skyblue")
plt.title("Histogram of Total Bill")
plt.show()

# 💡 Pro-Tip: The 'bins' parameter in histplot() is crucial. Too few bins oversimplify the data, while too many bins create noise.
#  Always experiment with the bin size!
```

### Here, we will create a KDE (Kernel Density Estimate) plot using sns.kdeplot() to visualize a smooth, continuous curve of the distribution.

```python
# Here, I am creating a KDE plot with a green fill to show the continuous probability density.

sns.kdeplot(tips["total_bill"], fill=True, color="green")
plt.title("KDE Plot of Total Bill")
plt.show()
```

### Here, we will create a Rug Plot using sns.rugplot() to draw a small vertical tick for every single actual data point.

```python
# Here, I am creating a rug plot on the x-axis to show the exact location of each data point.

sns.rugplot(x=tips["total_bill"], height=0.1, color="red")
plt.title("Rugplot: Total Bill Values")
plt.show()
```

### Here, we will combine multiple distributions by adding the 'hue' parameter, allowing us to compare different categories simultaneously.

```python
# Here, I am creating a combined histogram and KDE curve, separated by gender using the 'hue' parameter.

sns.histplot(x="total_bill", hue="sex", data=tips, kde=True, bins=20, palette="Set2")
plt.title("Histogram by Gender (with KDE)")
plt.show()

# 💡 Pro-Tip: Setting kde=True inside sns.histplot() is the most modern and efficient way to get both a histogram and a smooth density curve in a single plot.
#  You rarely need to use sns.kdeplot() separately anymore!
```

### Key takeaways

- histplot() → flexible histogram (can combine with KDE).

- kdeplot() → smooth density curve.

- rugplot() → quick look at raw values.

- distplot() is deprecated → prefer histplot().

- Use hue to compare groups in the same distribution plot.
