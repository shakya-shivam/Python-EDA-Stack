# Seaborn

## Here, we will explore the basics of Seaborn, a powerful library built on top of Matplotlib for creating beautiful statistical graphics.

* We will start by importing the required dependencies.

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

### 1. Built-in Datasets

* Here, we will load some built-in sample datasets (iris, tips, and penguins) using sns.load_dataset() to practice our visualizations.

```python
iris = sns.load_dataset('iris')
iris
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
penguins = sns.load_dataset('penguins')
penguins
```

### 2. Plots

* Here, we will create a Scatter Plot using sns.scatterplot() to visualize the relationship between total bill and tip amount.

```python
# Here, I am creating a scatter plot to show the correlation between the 'total_bill' and 'tip' columns.

sns.scatterplot(x="total_bill", y="tip", data=tips)
plt.title("Scatter Plot: Total Bill vs Tip")
plt.show()
```

* Here, we will create a Line Plot using sns.lineplot() to observe the trend of average tips across different days.

```python
# Here, I am creating a line plot to track the average 'tip' per 'day'. Seaborn automatically calculates the mean and shows the confidence interval.

sns.lineplot(x="day", y="tip", data=tips, estimator="mean", marker="o")
plt.title("Line Plot: Average Tip by Day")
plt.show()

# 💡 Pro-Tip: Unlike Matplotlib, Seaborn's lineplot() automatically aggregates multiple data points for the same x-value and 
# plots the mean (estimator="mean") along with a shaded 95% confidence interval.
```

* Here, we will customize the overall aesthetics of our plots using Seaborn's global styling and color palettes.

```python
# Here, I am setting the global background style to 'darkgrid'.

sns.set_style("darkgrid")        # options: white, dark, darkgrid, whitegrid, ticks


# Here, I am changing the default color palette to 'pastel'.

sns.set_palette("pastel")         # others: deep, bright, dark, muted, colorblind

sns.scatterplot(x="total_bill", y="tip", hue="time", data=tips, s=80)
plt.title("Styled Scatter with Hue & Palette")
plt.show()

# 💡 Pro-Tip: The 'hue' parameter is Seaborn's superpower! It automatically colors the data points based on a categorical
#  column (like 'time') and generates a ready-to-use legend without extra code.
```
