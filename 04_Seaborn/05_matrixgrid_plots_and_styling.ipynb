# Seaborn 

## Here, we will learn about Matrix Plots (like Heatmaps and Clustermaps) to visualize data in a color-coded grid format, and explore Seaborn's styling options.

* We will start by importing the required dependencies.

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Here, I am loading the built-in 'tips' dataset to practice matrix plots.

tips = sns.load_dataset('tips')
tips
```

### Here, we will use sns.heatmap() to visualize the correlation matrix, using colors to indicate the strength of relationships between numerical variables.

```python
# Here, I am calculating the correlation matrix and plotting it as a heatmap with annotations (annot=True).

corr = tips.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()

# 💡 Pro-Tip: Before creating a heatmap, you must have data in a matrix form. Using .corr() or 
#  .pivot_table() is the standard way to prepare data for a heatmap.
```

### Here, we will use sns.clustermap() to plot a matrix dataset as a hierarchically-clustered heatmap.

```python
# Here, I am creating a hierarchically clustered heatmap using the 'viridis' color map.

sns.clustermap(corr, annot=True, cmap="viridis")

# 💡 Pro-Tip: clustermap() automatically clusters similar rows and columns together using hierarchical clustering,
#  which is why the rows/columns might not be in their original order!
```

### Here, we will apply global styles and color themes to customize the overall look of our Seaborn plots.

```python
# Here, I am setting the global background style to 'darkgrid'.

sns.set_style("darkgrid")       # options: white, dark, darkgrid, whitegrid, ticks


# Here, I am setting the global color palette to 'Set2'.

sns.set_palette("Set2")


# Here, I am creating a scatter plot to demonstrate the newly applied global style and palette.

sns.scatterplot(x="total_bill", y="tip", hue="sex", data=tips, s=80)
plt.title("Styled Scatter with Darkgrid & Palette")
plt.show()
```
