# Seaborn

## Here, we will learn about Categorical Plots in Seaborn, which are essential for visualizing the relationship between a numerical variable and one or more categorical variables.

* We will start by importing the required dependencies.

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Here, I am loading the built-in 'tips' dataset to practice categorical plots.

tips = sns.load_dataset('tips')
tips
```

### Here, we will create a Bar Plot using sns.barplot() to estimate central tendency and visualize the average value across categories.

```python
# Here, I am creating a bar plot to show the average 'total_bill' for each 'day'.

sns.barplot(x="day", y="total_bill", data=tips)
plt.title("Barplot: Avg Total Bill by Day")
plt.show()

# 💡 Pro-Tip: By default, sns.barplot() calculates the mean (average). You can change this behavior using the 'estimator' parameter 
# (e.g., estimator=np.median or estimator=np.sum).
```

### Here, we will create a Count Plot using sns.countplot() to visualize the frequency or number of occurrences in each category.

```python
# Here, I am creating a count plot to display the total number of records (frequency) for each 'day'.

sns.countplot(x="day", data=tips)
plt.title("Countplot: Number of Records by Day")
plt.show()

# 💡 Pro-Tip: sns.countplot() does not accept a 'y' parameter because it automatically calculates the count on the y-axis 
# based on the categories provided in the 'x' parameter.
```

### Here, we will create a Box Plot using sns.boxplot() to visualize the distribution, median, quartiles, and outliers of our data.

```python
# Here, I am creating a box plot to analyze the spread and outliers of 'total_bill' across different days.

sns.boxplot(x="day", y="total_bill", data=tips)
plt.title("Boxplot: Total Bill by Day")
plt.show()
```

### Here, we will create a Violin Plot using sns.violinplot(), which combines a boxplot with a kernel density estimation to show the full distribution of the data.

```python
# Here, I am creating a violin plot to visualize the underlying probability distribution of 'total_bill' for each 'day'.

sns.violinplot(x="day", y="total_bill", data=tips, inner="box")
plt.title("Violinplot: Total Bill by Day")
plt.show()

# 💡 Pro-Tip: Swarm plots do not scale well to large datasets because the algorithm to avoid overlapping points becomes computationally heavy.
#  Use them only for relatively small datasets!
```

### Here, we will create a Strip Plot using sns.stripplot() to draw a scatterplot where one variable is categorical.

```python
# Here, I am creating a strip plot with jitter enabled to spread out overlapping data points.

sns.stripplot(x="day", y="total_bill", data=tips, jitter=True)
plt.title("Stripplot: Total Bill by Day")
plt.show()
```

### Here, we will create a Swarm Plot using sns.swarmplot() to draw a categorical scatterplot with non-overlapping points, providing a better view of the distribution.

```python
# Here, I am creating a swarm plot to ensure no data points overlap, giving a clear view of the distribution density.

sns.swarmplot(x="day", y="total_bill", data=tips)
plt.title("Swarmplot: Total Bill by Day")
plt.show()
```

| Plot        | Best for |
|-------------|-----------|
| **barplot** | Compare average values across categories |
| **countplot** | Show how many items fall in each category |
| **boxplot** | See median, quartiles, and outliers |
| **violinplot** | Same as boxplot + shows distribution |
| **stripplot** | View every observation (good for small data) |
| **swarmplot** | Like stripplot but avoids overlap |
