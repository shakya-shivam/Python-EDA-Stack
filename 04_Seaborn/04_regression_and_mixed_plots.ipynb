# Seaborn

## Here, we will learn about Regression and Mixed Plots in Seaborn to analyze trends and complex relationships between multiple variables.

* We will start by importing the required dependencies.

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Here, I am loading the built-in 'tips' dataset to practice regression plots.

tips = sns.load_dataset('tips')
tips
```

### Here, we will use sns.regplot() to plot data and automatically fit a linear regression model to show the trend.

```python
# Here, I am plotting 'total_bill' against 'tip' and adding a green regression line to show the trend.

sns.regplot(x="total_bill", y="tip", data=tips, color="green")
plt.title("regplot(): Bill vs Tip with Regression Line")
plt.show()

# 💡 Pro-Tip: regplot() is an Axes-level function, meaning it draws onto a specific existing Matplotlib axes. You can use it easily inside plt.subplot().
```

### Here, we will use sns.lmplot(), which is similar to regplot but allows us to split the regression lines across different categories (facets) using the 'hue' parameter.

```python
# Here, I am creating a regression plot where the data is split by 'sex' (using hue), creating separate trend lines for each group.

sns.lmplot(x="total_bill", y="tip", hue="sex", data=tips, height=5, aspect=1.2)
plt.title("lmplot(): Regression by Gender")
plt.show()

# 💡 Pro-Tip: lmplot() is a Figure-level function. It creates its own figure grid and cannot be easily placed inside a standard Matplotlib subplot grid.
```

### Here, we will use sns.jointplot() to draw a bivariate plot with marginal distributions (like scatter plots with histograms on the axes).

```python
# Here, I am creating a joint plot with a central scatter plot and histograms on the margins.

sns.jointplot(x="total_bill", y="tip", data=tips, kind="scatter")


# Here, I am creating a joint plot using KDE (Kernel Density Estimate) to show contours of density instead of a scatter plot.

sns.jointplot(x="total_bill", y="tip", data=tips, kind="kde", fill=True)
```

### Here, we will use sns.pairplot() to visualize pairwise relationships across the entire dataset in a grid format.

```python
# Here, I am loading the 'iris' dataset and creating a pairplot to compare all numerical variables against each other, color-coded by 'species'.

iris = sns.load_dataset("iris")
sns.pairplot(iris, hue="species")
plt.suptitle("Pairplot: Iris Dataset", y=1.02)
plt.show()

# 💡 Pro-Tip: pairplot() is fantastic for initial EDA, but avoid using it on massive datasets with dozens of columns, 
# as it will take a long time to render and become unreadable!
```
