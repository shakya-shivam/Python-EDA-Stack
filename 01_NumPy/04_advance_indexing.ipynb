# NumPy

## In this file, we will explore advanced concepts like fancy indexing, conditional selection, sorting, concatenation, and stacking in NumPy.

```python
import numpy as np
```

### 1. Fancy Indexing

* Fancy indexing allows us to extract specific, non-sequential elements from an array by passing an array of indices.

* It works efficiently on both 1-D and 2-D arrays.

```python
arr = np.array([10, 20, 30, 40, 50, 60]) # Here, I am creating and printing that array.
print("Original:", arr)
```

```python
# Pick elements at positions 0,2,4
print("Fancy indexing:", arr[[0, 2, 4]]) # Passing a list of indices [0, 2, 4] to extract specific elements from the array.
```

```python
matrix = np.arange(1, 13).reshape(3, 4) # Generating numbers 1 to 12 using np.arange() and reshaping them into a 3x4 2-D matrix.
print("Matrix:\n", matrix)
```

```python
# Here, I am extracting [2, 7] from matrix.
print("Select [row0,col1], [row1,col2]:", matrix[[0,1],[1,2]])
```

### 2. Conditional Selection

* Conditional selection allows us to filter or replace values based on specific boolean conditions.

```python
# Pro-Tip: np.where acts like a vectorized if-else statement, heavily used in Data Cleaning to replace outliers.
arr = np.array([5, 12, 18, 7, 30])

print(np.where(arr > 10))         # Indices where arr > 10
print(np.where(arr > 10, arr, 0)) # Replace values: if >10 keep, else put 0
```

### 3. Sorting and Ordering

* Exploring how to sort arrays and find the indices of sorted elements.

```python
arr = np.array([42, 5, 18, 23])
print("Sorted:", np.sort(arr))

# Pro-Tip: argsort is extremely useful in AI/ML to find the original index of the highest probability predictions.
idx = np.argsort(arr)
print("Indices:", idx)
print("Use indices:", arr[idx])
```

```python
mat = np.array([[3,7,1],[9,4,6]])
print(np.sort(mat, axis=1))   # sort each row
print(np.sort(mat, axis=0))   # sort each column
```

### 4. Concatenation

* Concatenation is used to join two or more arrays along an existing axis.

```python
a = np.array([1,2,3])
b = np.array([4,5,6])

print(np.concatenate([a, b]))
```

### 5. Stacking

* Stacking joins arrays along a new axis (vertically or horizontally), which is highly useful in Machine Learning for combining features.

```python
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])

# Very useful in Machine Learning for combining different feature columns into a single dataset.
print("vstack:\n", np.vstack((A,B)))   # vertical
print("hstack:\n", np.hstack((A,B)))   # horizontal
```
