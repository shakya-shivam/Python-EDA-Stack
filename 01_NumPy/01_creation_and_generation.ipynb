# Numpy

## In this file we will learn about NumPy basics.

### We will focus on some of the most important aspects of NumPy: vectors, arrays, matrices and number generation.

* First we will import all dependencies to use NumPy in this file.

```python
import numpy as np
```

* NumPy arrays essentially come in two flavors: vectors and matrices. Vectors are strictly 1-d arrays and matrices are 2-d (but you should note a matrix can still have only one row or one column).

### Creating NumPy arrays from a Python list

* We can create an array by directly converting a list or list of lists.

```python
a = [1,2,3,4,5]
arr = np.array(a) # Remember: NumPy arrays are strictly homogeneous (store only one data type).
#                 # If you mix data types, NumPy will use implicit upcasting/type coercion.
#                 # For example, mixing numbers and strings converts everything to Unicode Strings.
arr
```

* Let's create a matrix using list.

```python
l = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
arr = np.array(l) # In 2-d array we have Row and Column, vertical = column, horizontal = row.
arr
```

### Built in Methods

* Here, We will see some basic inbuilt methods.

```python
np.zeros(5) # To create a 1-D array, we specify the number of elements (e.g., how many zeros we want).
```

```python
np.zeros((4, 5)) # To create 2-d array, we will enter rows first then column.
```

```python
np.ones(5) # To create a 1-D array, we specify the number of elements (e.g., how many ones we want).
```

```python
np.ones((5, 4)) # To create 2-d array, we will enter rows first then column.
```

```python
np.full(5, 7) # To create a 1-D array filled with a specific number, enter the shape first, then the fill value.
```

```python
np.full((5, 5), 8) # To create 2-d array of any number in bracket enter row first and then column then enter that number.
```

```python
np.arange(1, 11, 1) # This method work same as Python range function.
```

```python
np.linspace(1, 10, 5) # The parameters are: start point, stop point, and the number of linearly spaced elements you want.
```

```python
np.eye(5) # This will give us a 2-d array with identity matrix.
```

```python
np.random.rand(5, 5) # This will give us a 2-d array of random elements. It provide values between 0 and 1 only.
```

```python
np.random.randint(1, 50, (4, 4)) # This gives a 2-D array of random integers. Parameters: start, stop (exclusive),
#                                # and shape (rows, columns).
```

```python
np.random.randn(2, 3)  # Provides random values from a Standard Normal Distribution (mean=0, variance=1).
```
