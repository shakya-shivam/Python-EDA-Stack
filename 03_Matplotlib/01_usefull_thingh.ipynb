# Matplotlib

## Here, we will learn about the basics of Matplotlib for creating static, interactive, and animated data visualizations in Python.

* We will start by importing the required dependencies. 

```python
import numpy as np
import matplotlib.pyplot as plt
```

### 1. Plotting Two Arrays

* Here, we will learn how to create a basic line plot using two simple arrays.

```python
# Here, I am defining two lists, x and y, to use as coordinates for the plot.

x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 40]

plt.plot(x, y)  


# Here, I am plotting the y values against the x values to create a basic line graph.

plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Basic Line Plot with Two Arrays")


# Here, I am rendering and displaying the final plot.

plt.show()
```

### 2. Multiple Plots on a Single Canvas

* Here, we will learn how to draw multiple lines on the same plot, add a legend, and customize the canvas size.

```python
# Here, I am defining one x-axis array and two different y-axis arrays (y1 and y2).

x = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]
y2 = [1, 2, 3, 4, 5]


# Here, I am setting the canvas size to 6x4 inches.

plt.figure(figsize=(6,4))   

plt.plot(x, y1, label="Squares", color="blue")
plt.plot(x, y2, label="Line", color="red")

plt.title("Multiple Plots on the Same Canvas")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True)

plt.show()

# 💡 Pro-Tip: Always call plt.legend() AFTER defining the 'label' argument inside plt.plot(). Otherwise, the legend will be empty!
```

### 3. Subplots

* Here, we will use plt.subplot() to create multiple separate plots within the same figure grid.

```python
# Here, I am defining one x-axis array and two different y-axis arrays (y1 and y2).

x = [1, 2, 3, 4, 5]
y1 = [2, 4, 6, 8, 10]
y2 = [1, 4, 9, 16, 25]


# Here, I am activating the first subplot position in a 2x2 grid structure.

plt.subplot(2, 2, 1)
plt.plot(x, y1, 'r--')
plt.title("Plot 1")


# Here, I am activating the fourth subplot position in the same 2x2 grid.

plt.subplot(2, 2, 4)
plt.plot(x, y2, 'g*')
plt.title("Plot 2")

plt.tight_layout()
plt.show()

# 💡 Pro-Tip: plt.tight_layout() is a lifesaver! It automatically adjusts subplot parameters so that axes labels and titles don't overlap.
```

### 4. Matplotlib Object-Oriented Interface

* Here, we will explore the Object-Oriented (OO) interface, which is the recommended approach for complex, highly customizable plots (using Figure and Axes objects).

```python
# Here, I am providing sample x and y data points.

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]


# Step 1: Here, I am creating the main Figure object, which acts as the blank canvas.

fig = plt.figure(figsize=(6,4))


# Step 2: Here, I am manually placing an Axes object on the canvas using relative dimensions [left, bottom, width, height].

ax = fig.add_axes([0.1, 0.1, 0.8, 0.8])


# Step 3: Here, I am plotting the line directly onto the specific Axes object I just created.

ax.plot(x, y, color="purple", marker="o", linestyle="-")


# Step 4: Here, I am customizing the specific Axes by setting its title, labels, and enabling the grid.

ax.set_title("Object-Oriented Plot")
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")
ax.grid(True)

plt.show()

# 💡 Pro-Tip: While plt.plot() is quick for simple charts, ALWAYS use the Object-Oriented approach (fig, ax = plt.subplots()) in production code for granular control over multiple axes.
```

### 5. Figure Size, Aspect Ratio & DPI in Matplotlib

* Here, we will learn how to control the physical size and image quality (DPI) of our plots.

```python
# Here, I am providing sample x and y data points.

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]


# Here, I am creating a figure with an 8x4 inch size and setting the resolution to 120 DPI (Dots Per Inch).

plt.figure(figsize=(8, 4), dpi=120)   

plt.plot(x, y, color="teal", marker="o")
plt.title("Custom Figure Size & DPI")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)

plt.show()
```

### 6. Special Plot Types: Scatter, Histogram & Boxplot

* Here, we will quickly explore different plot types available in Matplotlib before moving to Seaborn.

```python
# Here, I am setting up standard x and y lists, plus an array of 100 random normally distributed numbers.

x = [1, 2, 3, 4, 5]
y = [5, 7, 4, 6, 8]
nums = np.random.randn(100)

plt.figure(figsize=(12,4))


# Here, I am creating a scatter plot in the first subplot position to show individual data points.

plt.subplot(1,3,1)
plt.scatter(x, y, color="red")
plt.title("Scatter")


# Here, I am creating a histogram in the second subplot position to show the distribution of the random numbers across 10 bins.

plt.subplot(1,3,2)
plt.hist(nums, bins=10, color="skyblue")
plt.title("Histogram")


# Here, I am creating a boxplot in the third subplot position to visualize the spread and identify outliers in the random numbers.

plt.subplot(1,3,3)
plt.boxplot(nums)
plt.title("Boxplot")

plt.tight_layout()
plt.show()
```
